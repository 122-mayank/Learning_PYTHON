```python
fruits = ["Apple" , "Banana" ,"Cherry"]
print(fruits)
```

```python
for i in range(0,len(fruits)):
    print(fruits[i])
```

```python
for key in fruits:
    print(key)
```

```python
for i , val in enumerate(fruits):
    print(f"The index is {i} value with fruit {val}")
```

```python
my_set = set(fruits)
print(my_set)
```

```python
dict = {"A":12 , "B" : 23, "C":34}
for key , val in dict.items():
    print( key , val)
```

```python
numbers = [1, 2  , 34, 4 , 56]
square = [x ** 2 for x in numbers] #result for the variable in data structures
print(square)
```
