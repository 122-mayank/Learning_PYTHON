```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.array([2022 , 2023, 2024 , 2025])
y = np.array([125 , 150 , 175 , 200])
y1 = np.array([350 , 120 , 450 , 12])
plt.plot(x, y , marker ="o" , markersize = 6,markerfacecolor="#bd34eb",markeredgecolor="pink",label ="Class A")
plt.plot(x, y1 , marker ="o" , markersize = 6,markerfacecolor="#bd34eb",markeredgecolor="pink",label ="Class B")
plt.xlabel("Year of Placement")
plt.ylabel("No of students placed")
plt.title("Placement tracking graph")
plt.legend()
plt.show()
```

```python
xpoints = np.array([1 ,2 , 6,8  , 9])
ypoints = np.array([3 ,4 , 8,  9, 10])

plt.plot(xpoints , ypoints)
plt.show()
```

```python
"""
Default X-Points
If we do not specify the points on the x-axis, they will get the default values 0, 1, 2, 3 etc., depending on the length of the y-points.

So, if we take the same example as above, and leave out the x-points, the diagram will look like this:


"""

ypoints = np.array([3, 8, 1, 10, 5, 7])

plt.plot(ypoints)
plt.show()
```

```python
sim =np.array(["Jio","Airtel","VI"])
consumers = np.array([100000 , 10000, 1000])
plt.barh(sim , consumers , color ="blue")
plt.show()
```

```python
normal = np.random.normal(70 , 30 , 120)
plt.hist(normal , bins = 40)
plt.show()
```

```python
sim = np.array(["Jio","Airtel","VI","BSNL"])
consumers = np.array([450 ,219 , 290 ,80])
plt.pie(consumers , labels = sim, autopct ="%1.2f%%",explode =[0 , 0.2 , 0 , 0],startangle=120,shadow=0.6)
plt.savefig("piechart.png")
plt.show()
```

```python
#sactter plot
week_ex = np.array([1 ,2 ,3 ,4, 5 ,6 ,7, 8])
weight = np.array([75, 74, 76, 72 , 70 , 65 ,66 , 60])
plt.scatter(week_ex, weight, marker="*",color="pink",edgecolors="black")
plt.grid()
plt.show()
```

```python
#FMt
"""
Format Strings fmt
You can also use the shortcut string notation parameter to specify the marker.

This parameter is also called fmt, and is written with this syntax:

marker|line|color


Marker Size
You can use the keyword argument markersize or the shorter version, 
(ms) to set the size of the markers:

Marker Color
You can use the keyword argument markeredgecolor or the shorter (mec) 
to set the color of the edge of the markers:

You can use the keyword argument markerfacecolor or the shorter 
(mfc) to set the color inside the edge of the markers:

"""

ypoints = np.array([3 , 8 , 1, 10])
plt.plot(ypoints , 'o-g',ms = 20 , mec='b',mfc='r')
"""
marker is --> o
line is --> -
color is --> r
"""
plt.show()
```
